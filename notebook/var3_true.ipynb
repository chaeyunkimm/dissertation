**True VAR(3) Conditional Mean**

$$
\mathbf{X}_{t+1}
=
A_1\mathbf{X}_t
+
A_2\mathbf{X}_{t-1}
+
A_3\mathbf{X}_{t-2}
+
\boldsymbol{\varepsilon}_{t+1},
\qquad
\boldsymbol{\varepsilon}_{t+1}\sim N_3(\mathbf{0},\Sigma)
$$

$$
\mathbf{X}_{t+1}
\mid
\mathcal{F}_t
\sim
N_3
\left(
\boldsymbol{\mu}_{t+1},
\Sigma
\right)
$$

$$
\boxed{
\boldsymbol{\mu}_{t+1}
=
A_1\mathbf{X}_t
+
A_2\mathbf{X}_{t-1}
+
A_3\mathbf{X}_{t-2}
}
$$

```python
def true_var3_3d_conditional_mean(x_t, x_t_minus_1, x_t_minus_2, A1, A2, A3):

    x_t = np.asarray(x_t)
    x_t_minus_1 = np.asarray(x_t_minus_1)
    x_t_minus_2 = np.asarray(x_t_minus_2)

    mean_cond = ( A1 @ x_t + A2 @ x_t_minus_1 + A3 @ x_t_minus_2 )

    return mean_cond
```

X_{t+1}^{(j)}
\mid

\sim
N

**True VAR(3) Conditional Marginal PDF**

$$
p_{\mathrm{true}}^{(j)}
\left(
y_j\mid\mathcal{F}_t
\right)
=
\frac{1}{\sqrt{2\pi\Sigma_{jj}}}
\exp
\left[
-\frac{
\left(
y_j-\mu_{t+1}^{(j)}
\right)^2
}{
2\Sigma_{jj}
}
\right]
$$



```python
def true_var3_3d_conditional_marginal_pdf(y_j, j, mean_cond, Sigma):

    mean_cond = np.asarray(mean_cond)
    Sigma = np.asarray(Sigma)

    sd_j = np.sqrt(Sigma[j, j])

    p_cond_j = norm.pdf(y_j, loc=mean_cond[j], scale=sd_j)

    return p_cond_j
```

**True VAR(3) Conditional Marginal CDF**

$$
P_{\mathrm{true}}^{(j)}
\left(
y_j\mid\mathcal{F}_t
\right)
=
\Phi
\left(
\frac{
y_j-\mu_{t+1}^{(j)}
}{
\sqrt{\Sigma_{jj}}
}
\right)
$$

```python
def true_var3_3d_conditional_marginal_cdf(y_j, j, mean_cond, Sigma):

    mean_cond = np.asarray(mean_cond)
    Sigma = np.asarray(Sigma)

    sd_j = np.sqrt(Sigma[j, j])

    P_cond_j = norm.cdf(y_j, loc=mean_cond[j], scale=sd_j)

    return P_cond_j
```

**True VAR(3) Conditional PIT**

$$
U_{t+1,\mathrm{true}}^{(j)}
=
P_{\mathrm{true}}^{(j)}
\left(
X_{t+1}^{(j)}
\mid
\mathcal{F}_t
\right)
=
\Phi
\left(
\frac{
X_{t+1}^{(j)}
-
\mu_{t+1}^{(j)}
}{
\sqrt{\Sigma_{jj}}
}
\right)
$$

$$
\mathbf{U}_{t+1,\mathrm{true}}
=
\left(
U_{t+1,\mathrm{true}}^{(1)},
U_{t+1,\mathrm{true}}^{(2)},
U_{t+1,\mathrm{true}}^{(3)}
\right)
$$

```python
def true_var3_3d_conditional_pit(y, mean_cond, Sigma):

    y = np.asarray(y)
    d = len(y)

    U_true = np.zeros(d)

    for j in range(d):

        U_true[j] = true_var3_3d_conditional_marginal_cdf(
            y_j=y[j], j=j, mean_cond=mean_cond, Sigma=Sigma
        )

    return U_true
```

**True Gaussian Copula Correlation Matrix**

$$
D
=
\operatorname{diag}
\left(
\sqrt{\Sigma_{11}},
\sqrt{\Sigma_{22}},
\sqrt{\Sigma_{33}}
\right)
$$

$$
R_{\mathrm{true}}
=
D^{-1}\Sigma D^{-1}
$$

$$
\rho_{jk}
=
\frac{
\Sigma_{jk}
}{
\sqrt{\Sigma_{jj}\Sigma_{kk}}
}
$$

$$
R_{\mathrm{true}}
=
\begin{pmatrix}
1 & \rho_{12} & \rho_{13} \\
\rho_{12} & 1 & \rho_{23} \\
\rho_{13} & \rho_{23} & 1
\end{pmatrix}
$$

```python
def true_gaussian_copula_correlation(Sigma):

    Sigma = np.asarray(Sigma)

    sd = np.sqrt(np.diag(Sigma))
    R = Sigma / np.outer(sd, sd)

    return R
```
